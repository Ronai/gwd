## Consultar informações sobre eventos

O módulo `gwosc.datasets` fornece ferramentas para busca de conjuntos de dados, incluindo eventos, catálogos e versões completas de dados de linhagens.

Por exemplo, podemos buscar eventos no [GWTC-1 catalog](https://gwosc.org/eventapi/html/GWTC-1-confident/), o catálogo de todos os eventos das campanhas de observação O1 e O2. Uma lista dos catálogos disponíveis pode ser vista em [Event Portal](https://gwosc.org/eventapi)

```python
from gwosc.datasets import find_datasets
from gwosc import datasets

#-- Liste todos os catálogos disponíveis
print("List of available catalogs")
print(find_datasets(type="catalog"))
```

```python
#-- Liste todos os eventos de ondas gravitacionais do catálogo GWTC-1.
gwtc1 = datasets.find_datasets(type='events', catalog='GWTC-1-confident')
print('GWTC-1 events:', gwtc1)
print("")
```

Observe que o nome do evento segue o padrão _GWyymmdd-vx_, em que _x_ é a última versão disponível para o conjunto de dados fornecido pelo GWOSC.

```python
#-- Imprima todos os conjuntos de dados de strain intensa das campanhas de observação do LIGO/Virgo/KAGRA.
runs = find_datasets(type='run')
print('Large data sets:', runs)
```

Atenção: As campanhas de observação cujos nomes terminam em "Disc" não são campanhas reais, mas sim estruturas que adicionamos ao nosso banco de dados por motivos técnicos; elas estão associadas a detecções candidatas de ondas gravitacionais (GW) publicadas antes da divulgação dos dados da campanha de observação correspondente. Em nosso jargão, os artigos sobre essas detecções são chamados de "artigos de descoberta" (*discovery papers*), daí o sufixo "Disc" (por exemplo, O4b1Disc -> GW241011).

`datasets.find_datasets` também aceita os argumentos nomeados `segment` e `detector` para restringir os resultados com base no tempo GPS e no detector:

```python
#-- Palavras-chave de detector e segmentos limitam os resultados da pesquisa.
print(datasets.find_datasets(type='events', catalog='GWTC-1-confident', detector="L1", segment=(1164556817, 1187733618)))
```

Usando `gwosc.datasets.event_gps`, podemos consultar o tempo GPS de um evento específico (isso também funciona sem o número da versão):

```python
from gwosc.datasets import event_gps
gps = event_gps('GW190425')
print(gps)
```

<div class="alert alert-info">Todos esses horários são retornados no sistema de tempo GPS, que contabiliza o número de segundos decorridos desde o início da época GPS à meia-noite (00:00) de 6 de janeiro de 1980. O GWOSC fornece um <a href="https://gwosc.org/gps/">Conversor de tempo GPS</a> você pode usar para traduzir para data e hora, ou você pode usar <a href="https://gwpy.readthedocs.io/en/v3.0.14/time/"><code>gwpy.time</code></a>.</div>

```python
# Você também pode fazer o inverso.
from gwosc.datasets import event_at_gps
print(datasets.event_at_gps(1240215503))
```

Observe que o método `event_at_gps` busca eventos ocorridos em um intervalo de 1 segundo em relação ao horário GPS fornecido. Se nenhum evento for encontrado, ele gerará um erro.

Podemos consultar o intervalo de tempo GPS para uma campanha de observação:

```python
from gwosc.datasets import run_segment
print(run_segment('O1'))
```

```python
# e vice-versa, também neste caso
from gwosc.datasets import run_at_gps
print(run_at_gps(1240215503))
```

Agora podemos usar o que aprendemos com `run_segment` e `find_datasets` para visualizar apenas os eventos de alta confiança em O1:

```python
O1_events = datasets.find_datasets(type='events', catalog='GWTC-1-confident', segment=run_segment('O1'))
print(O1_events)
```

## Consultar arquivos de dados
O módulo `gwosc.locate` fornece uma função para encontrar as URLs de arquivos de dados associados a um determinado conjunto de dados.

Para conjuntos de dados de eventos, é possível obter a lista de URLs utilizando apenas o nome do evento:

```python
from gwosc.locate import get_event_urls
urls = get_event_urls('GW150914')
print(urls)
```

Por padrão, esta função retorna todos os arquivos associados a um determinado evento, o que não é particularmente útil. No entanto, podemos filtrar qualquer um deles usando argumentos de palavra-chave; por exemplo, para obter a URL do arquivo de 32 segundos referente ao detector LIGO-Livingston:

```python
urls = get_event_urls('GW150914', duration=32, detector='L1')
print(urls)
```

Os diferentes interferômetros são designados por um identificador curto:

- `'G1'` - GEO600
- `'H1'` - LIGO-Hanford
- `'L1'` - LIGO-Livingston
- `'V1'` - (Advanced) Virgo
- `'K1'` - KAGRA

##  Consulta filtrada por parâmetros de fusão

O módulo `query_events` de `gwosc.datasets` permite obter uma lista de eventos filtrada por alguns parâmetros, semelhante ao que é feito pela função `Query` do [event portal](https://gwosc.org/eventapi/html/query/). Uma lista de parâmetros disponíveis pode ser encontrada [here](https://gwosc.readthedocs.io/en/stable/reference/gwosc.datasets.query_events.html) ou usando `query_events?`.

Vamos ver como usar este módulo para descobrir quais eventos foram detectados com uma relação sinal-ruído (SNR) de rede entre 25 e 30:

```python
from gwosc.datasets import query_events
selection = query_events(select=["25 <= network-matched-filter-snr <= 30"])
# isso é equivalente a 
# query_events(select=["network-matched-filter-snr <= 30", "network-matched-filter-snr>= 25"])
print(selection)
```

Observe que este módulo fornecerá a lista de **todas as versões disponíveis** para todos os eventos que atendem aos parâmetros exigidos. Por exemplo, nesta consulta, o evento GW190814 aparece listado duas vezes porque duas versões desse evento satisfazem o critério de SNR entre 25 e 30.
